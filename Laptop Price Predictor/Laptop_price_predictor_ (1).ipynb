```python
#Linear regression
# in column trans we pass function one hot encoder

step1=ColumnTransformer(transformers=[
    ('col_tnf' ,OneHotEncoder(sparse_output=False,drop='first'),[0,1,2,4,5])
    ],remainder='passthrough')
#now in 2nd step apply linear reg
step2=LinearRegression()
#now merge two steps into one using pipeline : assemble multiple things together
pipe=Pipeline([('step1',step1),('step2',step2)])
pipe.fit(X_train,Y_train)
y_pred=pipe.predict(X_test)

print('R2 score',r2_score(Y_test,y_pred))
print('MAE',mean_absolute_error(Y_test,y_pred))
print('MSE',mean_squared_error(Y_test,y_pred))
```

```python
%%writefile app.py
import streamlit as st
import pickle
import numpy as np

df = pickle.load(open('df.pkl', 'rb'))  # rb: read block
pipe = pickle.load(open('pipe.pkl', 'rb'))
st.title("Laptop Price Predictor")

company = st.selectbox("Brand", df['Company'].unique(), index=4)
Type = st.selectbox("Type", df['TypeName'].unique(), index=1)
cpu = st.selectbox("Processor", df['Cpu'].unique(), index=0)
ram = st.selectbox("RAM(in GB)", [2, 4, 6, 8, 12, 16, 24, 32, 64, 128], index=8)
gpu = st.selectbox("GPU", df['Gpu'].unique(), index=0)
os = st.selectbox("Operating System", df['OpSys'].unique(), index=0)
weight = st.number_input("Weight of the laptop")
touchscreen = st.selectbox("Touchscreen", ["No", "Yes"], index=0)  # Fixed spelling
Ips = st.selectbox("IPS", ["No", "Yes"], index=0)
screen_size = st.number_input("Screen Size (in Inches, calculated diagonally)",
                               min_value=15.6, step=0.1, value=15.6)
resolutions = st.selectbox(
    "ScreenResolution",
    [
        "2560x1600", "1440x900", "1920x1080", "2880x1800",
        "1366x768", "2340x1440", "3200x1800", "1920x1200",
        "2256x1504", "3840x2160", "2160x1440", "2560x1440",
        "1600x900", "2736x1824", "2400x1600"
    ],
    index=2
)

if st.button("Predict Price"):
    ppi = None
    if touchscreen == 'Yes':
        touchscreen = 1
    else:
        touchscreen = 0
    if Ips == 'Yes':
        Ips = 1
    else:
        Ips = 0
    try:
        X_res, Y_res = map(int, resolutions.split('x'))  # Error-safe splitting
        ppi = ((X_res**2) + (Y_res**2))**0.5 / screen_size
    except ValueError:
        st.error("Invalid resolution format.")
    query = np.array([[company, Type, cpu, ram, gpu, os, weight, touchscreen, Ips, ppi]])
    try:
        op = np.exp(pipe.predict(query))
        st.subheader(f"The predicted price of the laptop with the above configuration is {round(op[0])}")
    except Exception as e:
        st.error(f"Prediction failed: {str(e)}")
```
