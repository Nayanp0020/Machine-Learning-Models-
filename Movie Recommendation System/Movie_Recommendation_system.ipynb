```python
import ast
def convert(obj):# obj is an list we pass in def function i.e convert
  L=[] #it iterates when i is an dict i.e name and another is value
  for i in ast.literal_eval(obj):
    L.append(i['name'])
  return L
```

```python
# convert str to list using ast literal function
convert('[{"id": 12, "name": "Adventure"}, {"id": 878, "name": "Science Fiction"}, {"id": 28, "name": "Action"}]')
```

```python
def convert3(obj):
  L=[]
  counter=0
  for i in ast.literal_eval(obj):
    if counter !=3:
      L.append(i['name'])
      counter+=1
    else:
      break
    return L
```

```python
def fetch_director(obj):
  L=[]
  for i in ast.literal_eval(obj):
    if i['job']=='Director':
      L.append(i['name'])
      break
  return L
```

```python
#stemming " is the process of reduce inflected words to their root words"
# it ia a nlp technique that is used to reduce words to their base form
```

```python
['love','loved','loving']
['love','love','love']
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
ps.stem('loved') #gives back og word by removing affeclations
```

```python
def stemming(text):
  y=[]
  for i in text.split():
    y.append(ps.stem(i))
  return " ".join(y)
```

```python
stemming("love loved loving") #changes into root word i.e og word
```

```python
stemming("dance dancing danced") #root word is danc not dance ac to steming prop
```

```python
def recommend(movie):
  index=new_df[new_df['title']==movie].index[0]
  distances=sorted(list(enumerate(similarity[index])),
       key=lambda x:x[1],reverse=True)[1:11]
  for i in distances[1:11]:
    print(new_df.iloc[i[0]].title)
```

```python
%%writefile app.py
import streamlit as st
import pickle
import requests

movies=pickle.load(open('movies_list.pkl','rb'))
similarity=pickle.load(open('similarity.pkl','rb'))

def recommend(movie):
  index=movies[movies['title']==movie].index[0]
  distances=sorted(list(enumerate(similarity[index])),
       key=lambda x:x[1],reverse=True)[1:11]

  recommended_movie_names=[]

  for i in distances[1:11]:

    recommended_movie_names.append(movies.iloc[i[0]].title)
  return recommended_movie_names

st.title("Movie Recommender System")
movie_list=movies['title'].values
selected_movie=st.selectbox("Type or select a movie from the dropdown",movie_list)

if st.button('Show Recommendation'):
  recommended_movie_names=recommend(selected_movie)
  for i in recommended_movie_names:
    st.write(i)
  #c1,c2,c3=st.tabs(["1","2","3"])/columns()
  #with c1:
    #st.text(recommended_movie_names[0])
  #with c2:
    #st.text(recommended_movie_names[1])
  #with c3:
    #st.text(recommended_movie_names[2])
```
