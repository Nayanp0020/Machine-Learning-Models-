```python
img=cv2.imread("/content/p9.jpg")

model=cv2.CascadeClassifier("/content/haarcascade_frontalface_default.xml")
faces=model.detectMultiScale(img,1.1,8)  #this module is for face


model_glasses=cv2.CascadeClassifier("/content/haarcascade_eye_tree_eyeglasses.xml.1")
glasses=model_glasses.detectMultiScale(img,1.1,3) #this is for glasses detection

model_smile=cv2.CascadeClassifier("/content/haarcascade_smile.xml")
smile=model_smile.detectMultiScale(img,1.1,110) #this is for smile detection



for x,y,w,h in faces:
  cv2.rectangle(img,(x,y),(x+w,y+h),(255,0,0),2)

for x,y,w,h in glasses:
  cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),2)

for x,y,w,h in smile:
  cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),2)

cv2_imshow(img)

#face , eye from glasses, and smile detection
```
