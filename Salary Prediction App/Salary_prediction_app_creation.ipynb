```python
%%writefile app.py
import streamlit as st
import joblib
st.header("Salary predictor using linear regression")
st.caption("This Ml model was designed with a sample data obtained from kaggle . Linear regression was the algo used here")

model=joblib.load("lr_model")

experience_options = [round(i * 0.1, 1) for i in range(0, 131)]

years = st.selectbox("Select the number of years of experience", experience_options, index=45)

if st.button("PREDICT"):
    op = model.predict([[years]])
    st.subheader(f"The estimated salary for a person having {years} years of experience is ₹ {round(op[0])}")
```
